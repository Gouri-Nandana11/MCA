```
from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
import tensorflow as tf

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

train_images = train_images.reshape(
    (train_images.shape[0], 28 * 28)).astype('float32') / 255
test_images = test_images.reshape(
    (test_images.shape[0], 28 * 28)).astype('float32') / 255

model = Sequential([
    Dense(128, activation='relu', input_shape=(28 * 28,)),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy', metrics=['accuracy'])

model.fit(train_images, train_labels, epochs=5)

test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc}")
```

```
import polars as pl

# Create DataFrame
df = pl.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [23, 30, 27, 22],
    "Marks": [85, 92, 78, 88]
})

# Display data
print(df)

# Select columns
print(df.select(["Name", "Marks"]))

# Filter students with marks > 80
print(df.filter(pl.col("Marks") > 80))

# Add bonus marks
df = df.with_columns(
    (pl.col("Marks") + 5).alias("Bonus")
)

# Sort by marks
print(df.sort("Marks", descending=True))

# Average marks
avg = df.select(
    pl.col("Marks").mean().alias("Average Marks")
)

print(avg)
```

```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from pandas.plotting import parallel_coordinates

# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add species column
df["species"] = iris.target_names[iris.target]

# Create parallel coordinates chart
parallel_coordinates(
    df,
    "species"
)

# Add title
plt.title("Parallel Coordinates Chart - Iris Dataset")
plt.xlabel("Features")
plt.ylabel("Values")

plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Iris dataset
iris = sns.load_dataset("iris")

# Select numerical columns
features = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

# Calculate overall mean
overall_mean = iris[features].mean()

# Calculate mean for each species
species_mean = iris.groupby("species")[features].mean()

# Calculate deviation from overall mean
deviation = species_mean - overall_mean

# Plot deviation chart
deviation.T.plot(kind="bar", figsize=(10, 6))

plt.axhline(0, color="black", linewidth=1)
plt.title("Deviation Chart - Iris Dataset")
plt.xlabel("Features")
plt.ylabel("Deviation from Overall Mean")
plt.xticks(rotation=45)
plt.legend(title="Species")

plt.show()
```
